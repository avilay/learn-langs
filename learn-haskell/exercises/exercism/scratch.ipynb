```python
def rotate(offset: int, plain: str) -> str:
    chars = []
    for char in plain:
        base = ord("a") if char.islower() else ord("A")
        pos = ord(char) - base
        new_pos = (pos + offset) % 26
        new_char = chr(new_pos + base)
        chars.append(new_char)
    return "".join(chars)
```

```python
rotate(5, "omg")
```

```python
rotate(5, "OMG")
```

```python
def is_pangram(text: str) -> bool:
    letters = set([char.lower() for char in text if char.isalpha()])
    alphabets = set(string.ascii_lowercase)
    return alphabets.issubset(letters)
```

```python
is_pangram("the quick brown fox jumpes over the lazy dog.")
```

```python
is_pangram("helLLOW! ?World//")
```

```python
def is_loud(text: str) -> bool:
    return text == text.upper()


def is_question(text: str) -> bool:
    return text[-1] == "?"


def is_loud_question(text: str) -> bool:
    return is_loud(text) and is_question(text)


def response_for(text: str) -> str:
    text = text.strip()
    if not text:
        return "Fine. Be that way!"
    if is_loud_question(text):
        return "Calm down, I know what I'm doing!"
    if is_loud(text):
        return "Whoa, chill out!"
    if is_question(text):
        return "Sure."
    return "Watever."
```

```python
response_for("How are you?")
```

```python
response_for("HOW ARE YOU?")
```

```python
response_for("HELLO")
```

```python
response_for("   \n   \t  \r")
```

```python
def _collatz(steps: int, n: int) -> int:
    if n == 1:
        return steps
    if n % 2 == 0:
        return _collatz(steps + 1, n // 2)
    else:
        return _collatz(steps + 1, 3 * n + 1)


def collatz(n: int) -> int | None:
    if n <= 0:
        return None
    return _collatz(0, n)
```
