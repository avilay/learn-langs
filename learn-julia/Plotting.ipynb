  * `PlotlyJS`

```julia
using Pkg
```

```julia
Pkg.activate("ai"; shared=true)
```

```julia
using Plots
```

```julia
x = -3:0.1:3
```

```julia
f(v) = v^2
```

```julia
y = f.(x)
```

```julia
# Use the GR backend
gr()
```

```julia
plot(x, y, label="line")
scatter!(x, y, label="points")
```

The call to `scatter` has a `!` after it because I want it to modify the existing plot. Without it, `scatter` will create a new plot.

```julia
xlabel!("Number of pirates (approx)")
ylabel!("Global temparature (C)")
title!("Influence of pirate population on global warming.")
```

```julia
# Use the plotlyJS backend
plotlyjs()
```
