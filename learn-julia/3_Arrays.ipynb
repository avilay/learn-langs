## Declaration

### Syntax and Types

General syntax seems to be -
  * Commas are used for enumerating elements
  * Semicolons and newlines are used for stacking elements vertically
  * Spaces and double semicolons are used for stacking elements horizontally
  
$\mathcal{Vector}$ are used for 1D *column* vectors. Either commas or semicolons can be used to declare these. If I want to create a *row* vector, I'll have to use spaces. Furthermore I'll end up with a type of $1 \times 2$ $\mathcal{Matrix}$.

$\mathcal{Matrix}$ are used for 2D list of numbers. Here spaces are used for defining a row, and rows are separated (or stacked on top of each other) by semicolons. If I try to seperate a row by commas instead, I'll end up with the outer type being a vector.

$\mathcal{Array}$ are used for multi-dimensional tensors.

It is not possible to define jagged tensors.

The repr of a variable has the shape of the data as well as its type in there. E.g., when I see `3-element Vector{Int64}`, it means that the `typeof()` function will give me `Vector{Int64}`.

## Concatenation

The main function to use is [`cat(A...; dims)`](https://docs.julialang.org/en/v1/base/arrays/#Base.cat). There are a bunch of convenience functions like `vcat`, `hcat`, etc. But the easiest way to concat is to use a combination of newlines/semicolons and spaces/double-semicolons for this. Use as many semicolons as the dimension number I want to concatenate on, e.g., to concatenate on the third dimension use `;;;` and so on. A single semicolon will concatenate on the first dimension, i.e., it will stack rows on top of each other. This is the same behavior as that of newlines. A double semicolon will concatenate on the second dimension, i.e., it will stack columns next to each other. This is the same as using spaces.

````julia
# Column vector
x = [1, 2, 3]
````

````julia
# Will stack two xs on top of each other
[x; x]
````

````julia
# This will also stack two xs on top of each other
[
    x
    x
]
````

````julia
# Will stack two xs next to each other
[x x]
````

````julia
# This will also stack two xs next to each other
[x;; x]
````

````julia
B = [
    9 11
    10 12
]
````

````julia
# This will stack A on top of B
cat(A, B, dims=3)
````

````julia
[A;;; B]
````

I might think that `[A;B]` will stack all of `A` on top of `B`, but that is not true. It will try to concatenate on the first dimension. And in order to do that, all the other dimensions must match, i.e., $A \in \mathcal R^{2 \times 2 \times 2}$ and $B \in \mathcal R^{2 \times 2}$ or equivalently $B \in \mathcal R^{2 \times 2 \times 1}$. dim-1 does not need to match, that is the one that is going to be concatenated. dim-2 does match but dim-3 does not. So this operation will fail. The reason concatenation on the third dimension succeeds is because the first two dims match. 

Lets reshape `B` so that the last two dims match `A`. Note that because Julia is column-major, the columns are split into different dimensions. So if I was expecting the output to be -

$$
\begin{bmatrix}
\begin{bmatrix}
1 & 3 \\
2 & 4 \\
9 & 11
\end{bmatrix} \\
\begin{bmatrix}
5 & 7 \\
6 & 8 \\
10 & 12
\end{bmatrix}
\end{bmatrix}
$$

That is NOT GOING TO HAPPEN!

Reshaping `B` will split its columns into different dimensions, so I'll get the column elements stacked under the rows of each block of `A`.

````julia
# Lets reshape B so the last two dims match A
C = reshape(B, 1, 2, 2)
````

````julia
# Now everything works!
[A; C]
````

If I want to concatenate on the second dimension, i.e., I want to stack the elements of `B` horizontally onto each block of `A`, i.e., I want the output -

$$
\begin{bmatrix}
\begin{bmatrix}
1 & 3 & 9 \\
2 & 4 & 10 \\
\end{bmatrix} \\
\begin{bmatrix}
5 & 7 & 11 \\
6 & 8 & 12 \\
\end{bmatrix}
\end{bmatrix}
$$

Here again `B`'s columns are split into different dimensions, so each original column is stacked horinzontally to the columns in each block of `A`.

````julia
D = reshape(B, 2, 1, 2)
````

````julia
[A;; D]
````

## Indexing

Julia stores its multidim arrays in column major form, as opposed to numpy which uses the row-major form. There are two main ways to index into an array -
  * Cartesian indexing: Here I have to specify a value for each dimension. Each value can be a scalar, array, range, booleans, or `:`.
  * Linear indexing: Here I specify a single value. This single value can be a scalar or another array.

### Linear Indexing

Conceptually, the array is first unravled in its col-major form, and the elements provided as the index are chosen. For example if I say `A[3]`, it will give me the third element in the unravled array. If I say `A[[3, 50, 43]]` it will give me another array that looks like `[A[3], A[50], A[43]]`. The output will take the shape of the input index.

````julia
A = reshape(1:(2*3*3*2*2), 2, 3, 3, 2, 2)
````

````julia
A[[3, 50, 43]]
````

````julia
A[[3 50; 43 10]]
````

### Cartesian Indexing
This is the more familiar form of indexing. Unlike numpy where `A[1]` is shortcut for `A[1,:]`, in Julia I have to specify each and every dimension with a few exceptions.

````julia
# Using all scalars
A[1, 2, 2, 2, 2]
````

In the below example where my index is `[2, 2:3, 2, 2, 2]`, to begin with I ignore the first two dimensions and focus on which blocks to select. So I choose the `[:, :, 2, 2, 2]` block which is -
$$
\begin{matrix}
61 & 63 & 65 \\
62 & 64 & 66
\end{matrix}
$$

Then looking at the first two dims I know that I want the $2^{nd}$ row and all columns ranging from $[2, 3]$, which is just the $2^{nd}$ and $3^{rd}$ columns, which gives me the elements $64$ and $66$.

````julia
A[2, 2:3, 2, 2, 2]
````

In the below example `[:, 3, [1, 3], 2, 2]` one of the dimensions is specified as an array, but this does not mean Julia is going to use linear indexing here. The array is just a way to cherry-pick elements from that dimension. As before lets ignore the first two dimensions. I need the `[:, :, 1, 2, 2]` and `[:, :, 3, 2, 2]` blocks -

$$
\begin{bmatrix}
55 & 57 & 59 \\
56 & 58 & 60 \\
\end{bmatrix}
$$

and 

$$
\begin{bmatrix}
67 & 69 & 71 \\
68 & 70 & 72
\end{bmatrix}
$$

From here I need all the rows and the third column from each -

$$
\begin{bmatrix}
59 \\
60
\end{bmatrix}
$$

and 

$$
\begin{bmatrix}
71 \\
72
\end{bmatrix}
$$

There are two arrays in the index, so they will determine the shape of the output. The first one is `:` which is $2 \times 1$ and the second one is `[1, 3]` which is also $2 \times 1$, so the output will be $2 \times 2$. I admit I don't fully understand how the output is arranged in my pre-frontal cortex, but I have an intuitive feel for it, and for the time being that will have to do.

````julia
A[:, 3, [1, 3], 2, 2]
````

If I specify the 3rd dimension as `[1 3]` instead, this will give me a $1 \times 2$ that will need to be combined with $2 \times 1$ from `:` resulting in $2 \times 1 \times 2$

````julia
A[:, 3, [1 3], 2, 2]
````

This is more clear if I get rid of the one of the arrays in the index. 

````julia
A[1, 3, [1 3], 2, 2]
````

````julia
A[1, 3, [1, 3], 2, 2]
````

````julia
A[:, 3, 1, 2, 2]
````

The array inside the index can be multidim. The logic remains the same, blocks are cherry picked based on the indices in the index-array and then arranged based on the index-array's shape.

````julia
A[1, 3, [1 3; 2 3], 2, 2]
````

````julia
A[:, 3, [1 3; 2 3], 2, 2]
````

## Broadcasting

Broadcasting will apply a binary function (function that takes in two inputs) to each element of the 2 input arrays. This means that the arrays need to have the exact same shape, i.e., the number of dims should be the same, and the size of each dim should also be the same. If this is not the case, the `broadcast` function will try to replicate the values in a "sensible" manner to make the shapes the same. It works very similar to Python's broadcast, except the broadcasting replication blocks are from the left and not the right like in Python. For example if I need to broadcast a $2 \times 3 \times 1 \times 2 \times 2$ with a $2 \times 3 \times 3 \times 2 \times 2$ tensor, then main $2 \times 3$ matrix will be replicated thrice inside each of the four blocks.

Even though there is an explicit function called `broadcast`, people usually use the `.` operator syntactic sugar for this.

````julia
o = ones(Int, 2, 3, 3, 2, 2);
````

````julia
a = reshape(Vector(1:(2*3*1*2*2)), 2, 3, 1, 2, 2)
````

Now if I broadcast `a` with `o`, then I expect to see the "main" matrices replicated thrice.

````julia
o .* a
````

If the last dimension is missing, or is $1$, which all the same to Julia, the tensor will be "unsqueezed" in the last dimension and the entire outer block, which is the entirity of the tensor, will be replicated.

````julia
a = reshape(Vector(1:(2*3*3*2)), 2, 3, 3, 2)
````

````julia
o .* a
````

If the second dimension is $1$, i.e., columns are missing in each matrix, the existing column will be replicated in all the blocks. This is in keeping with the rule that everything to the left of the missing dimension is replicated.

````julia
b = reshape(Vector(1:(2*1*3*2*2)), 2, 1, 3, 2, 2)
````

````julia
o .* b
````

Again, just like numpy broadcasting, if any other dimension is missing, Julia cannot unsqueeze dimensions anywhere in the middle. `o` is $2 \times 3 \times 3 \times 2 \times 2$ and lets have another tensor `b_` that is $2 \times 3 \times 2 \times 2$. A new dimension can be added in the second place or the third place, but Julia does not know where. So this will fail.

````julia
b_ = reshape(b, 2, 3, 2, 2)
````

````julia
# This will not work
o .* b_
````

Just like in numpy, the mismatched dimension must be exactly $1$ for broadcast to work. If I try to broadcast $2 \times 2 \times 3 \times 2 \times 2$ with $2 \times 3 \times 3 \times 2 \times 2$, it will fail

````julia
c = reshape(Vector(1:(2*2*3*2*2)), 2, 2, 3, 2, 2)
````

````julia
# This will fail
o .* c
````

If the first dimension is $1$, then the rule of everything to the left will be replicated does not hold. In this case the row will be replicated.

````julia
d = reshape(Vector(1:(1*3*3*2*2)), 1, 3, 3, 2, 2)
````

````julia
o .* d
````

````julia
# A column vector and matrix will work with .+ but not with +
A = [
    1 2 3;
    4 5 6
]
v = [7, 8]
A .+ v
````

````julia
# Here I am not broadcasting. The add function expects the input arrays to be of the same size.
A + v
````

````julia
# A row vector and matrix will also work. Julia is able to figure out the right thing to do based on dimensions.
u = [7 8 9]
A .+ u
````

````julia
A + u
````

````julia
# If B is a matrix with the same shape as A, then both + and .+ will work and they will give the same result
B = [
    7 8 9;
    10 11 12
]
A + B
````

````julia
A .+ B
````

````julia
# Matrix-Matrix multiplication will work when the dims are (m, n) and (n, p) using the * operator. 
# But .* will not work in this case.
# .* will work with (m, n) and (m, n) matrix.
C = [
    7 8;
    9 10;
    11 12
]
A * C
````

## Linear Algebra

Julia has a bunch of other [linalg operators](https://docs.julialang.org/en/v1/stdlib/LinearAlgebra/).

  * `'` will give the transpose of a real valued matrix and the Hermitian (conjugate transpose) of a complex matrix.
  * `adjoint` and `'` are equivalent.

````julia
A = [
    1 2 3;
    4 5 6
]
A'
````

````julia
C = [
    1+0.1im 2+0.2im 3+0.3im;
    4+0.4im 5+0.5im 6+0.6im
]
C'
````

````julia
adjoint(C)
````

````julia
adjoint(A)
````
