````julia
function Base.show(io::IO, hist::Histogram)
    edgevals = collect(hist.edges[1])
    bins = collect(zip(edgevals, edgevals[2:end]))
    for ((lower, upper), freq) in zip(bins, h.weights)
        print(io, "[$lower, $upper): $freq\n")
    end
end
````

## Uniform Distribution

The [`rand`](https://docs.julialang.org/en/v1/stdlib/Random/#Base.rand) function is pretty versatile. It generates a uniform distribution over its inputs and outputs as many numbers in whatever shape I want. Its signature is -

```julia
rand([rng], [S], [dims...])
```

All its arguments are optional.

  * `rng`: I can pass in my own random number generator if I so choose, but for most cases I can ignore this param.
  * `S`: This is a very versatile param and the most confusing. This can be a collection from which to choose the sample, or it can be the type of value I want. By default `rand` will output `Float64`.
  * `dims`: The dimensions of the output. This can be a straightforward varargs, or it can be a tuple. However, if it is a tuple there are some confusing rules about whether the tuple will be interpreted as `S` or `dims`. So IMHO it is best to just supply these as varargs.

There is a similar function called `rand!` which will populate the passed in array but I have not covered it in this notebook.

````julia
# Generate a single random number in the interval [0, 1)
rand()
````

````julia
# Generate 100 random numbers in the interval [0, 1)
nums = rand(100)
````

````julia
h = fit(Histogram, rand(100))
````

We can see that these 100 values are uniformly distributed between $[0.0, 1)$.

````julia
typeof(h)
````

````julia
# S=1:10 and dims=100
nums = rand(1:10, 100)
````

````julia
countmap(nums)
````

Again, the integers are distributed uniformly in the interval `[1:10]`.

````julia
# S=1:10, dims=3,5
rand(1:10, 3, 5)
````

````julia
rand(3, 5)
````

````julia
# I can sample from a collection of any objects!
# The sampling will happen with replacement.
stocks = Dict(
    "GOOG" => 310.52,
    "AMZN" => 226.19,
    "APPL" => 278.28,
    "MSFT" => 478.53,
    "META" => 644.23,
    "NVDA" => 175.02,
    "TSLA" => 458.96
)
rand(stocks, 3)
````

## Range

[`range()`](https://docs.julialang.org/en/v1/base/math/#Base.range) is the workhourse function for generating a deterministic sequence of numbers. There are a number of methods with different ways of passing in the parameters, some of them can be passed in as keywords some as positional. Given my early stage of learning, I'll stick to passing all the parameters as keywords. Here are the different ways I can call `range()` -
  * Pass in any three of `start`, `step`, `stop`, or `length`.
  * Pass in two of `start`, `stop`, or `length`. The value of `step` will be $1$.
  * Pass in one of `stop` or `length`. Values of both `start` and `step` will be $1$.

This function returns a range style object that can be iterated over. I can also `collect()` all the values in a `Vector` if I want.

````julia
r = range(start=10, step=2, stop=20)
````

````julia
typeof(r)
````

````julia
for i in r
    print(i, " ")
end
````

````julia
collect(r)
````

````julia
# I can also use this syntactic sugar
for i in 10:2:20
    print(i, " ")
end
````

````julia
# This is similar to linspace
for i in range(start=10, stop=15, length=20)
    print(i, " ")
end
````
