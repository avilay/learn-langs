# Collections

## Built-In Functions

Ref: https://docs.julialang.org/en/v1/base/collections/

**General collections:**
  * `isempty(itr)`: check if the collection/iterator is empty or not.
  * `isdone(itr)`: sames as `isempty()` but for stateful iterators without consuming a value. Most iterators implement `isempty` using `isdone` anyway, so probably not very useful unless I am implementing my own iterator.
  * `empty!(itr)`: will clear out all the contents of the collection.
  * `length(itr)`: The number of elements remaining in the collection.

**Iterable collections:**
  * `in(item, itr)` or $x \in s$ and $x \notin s$: checks whether the given item is or is not in the given collection or not.
  * `eltype(itr)`: returns the type of elements in the iterator.
  * `unique(itr)`: returns vector of distinct elements in the iterator. It can do much more complicated things with multidimensional arrays, and function params, etc.
  * `allunique(itr)`: checks whether all the elements in the iterator are unique.
  * `allequal(itr)`: checks if all the elements in the iterator are equal to each other.
  * `reduce(op, itr; [init])`: `op` is a binary function. `init` is used for empty collections but according to the documentation "It is unspecified whether init is used for non-empty collections.". I guess it is dependent on the method implemented for the specific type. The associativity is also implementation dependent. There is a also a `reduce` that acts on arrays.
  * `foldl(op, itr; [init])`: Like `reduce` but guaranteed left associativity.
  * `foldr(op, itr; [init])`: Like `reduce` but guaranteed right associativity. 
  * `maximum()`, `minimum()`, `exterma()`: All of these have the signature `func(f, A::AbstractArray; dims)`. If I don't pass in a custom function and only pass in a vector, they work as expected. Even for arrays, it will find the maximum/minimum value(s) along the given dimension.
  * `argmax()`, `argmin()`: Signature `func(A; dims)`. Gets the index of the maximum or minimum values along the specified dims.
  * `findmax()`, `findmin()`: Signature `func(A; dims)`. Mixture of `maximun` and `argmax`, i.e., gets both the maximum/minimum values and their respective indexes.
  * `sum()`, `prod()`: Signature `func(A; dims)`.
  * `any([p,] itr)`, `all([p,] itr)`: Check if any or all items are true. `missing` values are also handled.
  * `count()`: If given a function param and an array (with dims) will count the number of elements for which `f` returns `true`. If given two strings, will count the number of occurences of the first string in the second string.
  * `foreach(f, c...)`: Call function `f` on each element of `c`. The result of the function is discarded, so typically `f` is something with side-effects.
  * `map(f, c...)`: There are other variants like `mapreduce`, `mapfoldl`, `mapfoldr` which first map and then reduce.
  * `first(itr, n)` and `last(itr, n)`: Gets the first (if `n` is not given) or the first n items. Similar to `xs[1:n]` but useful if the iterator is not indexable I guess. Similarly for `last`.
  * `front(tpl)` and `tail(tpl)`: Gets all but the last component and all but the first component.
  * `collect(itr)`: Collects all the elements of the iterator into a vector. When used with comprehensions, no need to explicitly call collect.
  * `filter(f, itr)`: Also works with arrays and dicts. For dicts, `f` is passed a `Pair`.
  * `replace()`: I can specify the replacements as pairs `replace([1, 2, 1, 3], 1=>0, 2=>4)` or I can specify a function that will replace each element with the output of the function. How is this different from `map`?

## Additional Utilities

Ref: https://docs.julialang.org/en/v1/base/iterators/

  * `zip(iters...)`
  * `enumerate(iter)`: Yields `(i, x)` where `i` is a counter starting at 1. It is not neccessarily a valid index especially for iterators whose indexes don't start with 1.
  * `rest(itr, state)`: Starts the iterator from the given state, not neccessarily from the beginning.
  * `countfrom(start=1, step=1)`: Iterator that counts forever.
  * `take(iter, n)`: Yield at most the first n elements of the iterator.
  * `takewhile(pred, iter)`: Yield elements as long as predicate is true.
  * `drop(iter, n)` and `dropwhile(pred, iter)`: Similar to `take`.
  * `cycle(iter[, n])`: Cycles through the iter forever or `n` times.
  * `repeated(x[, n])`: Generates `x` forever or `n` times.
  * `product(iters...)`: Cross product over all the iterators. Returns an ndarray. The first iterator changes the fastest.
  * `flatten((iter...))`: Argument iterators are concatenated. Note, the input is a single iterator that is made up of other iterators that I want to flatten.
  * `partition(col, n)`: Chunk the input collection in items of n elements.

```julia
a = 1:5
b = ["hello", "namaste", "hola", "halo"]
for (i, j) in zip(a, b)
    println("i=$i, j=$j")
end
```

```julia
mag7 = ["GOOG", "AMZN", "APPL", "MSFT", "META", "NVDA", "TSLA"]
for (idx, stock) in enumerate(mag7)
    println("[$idx] $stock")
end
```

```julia
for i in Iterators.rest([1, 2, 3, 4], 2)
    print(i, " ")
end
```

```julia
for i in Iterators.take(1:100, 3)
    print(i, " ")
end
```

```julia
names = ["Avilay", "Parekh", "APTG", "Lala", "Bhai"]
for i in Iterators.takewhile(name -> startswith(name, 'A'), names)
    print(i, " ")
end
```
