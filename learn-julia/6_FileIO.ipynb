# File IO

## Read/Write Primitives

Here are the simplified signatures of the read/write primitives. There are two builtin IO streams `stdin` and `stdout` -
  * `write(io, content)` -> Writes the content to the given IO stream. Returns the number of bytes written. The content is interpreted as a raw array of bytes, no encoding/decoding is done.
  * `write(filename, content)` -> Writes content to the given filename. It is supposed to write content as raw bytes, but somehow the file shows up ok.
  * `read(io, Type)` -> Returns the contents of the given IO stream in an object of the specified type. 
  * `read!(io, buf)` -> Reads the contents of the given IO stream into the given buffer. I think the buffer should be a bytearray, i.e., `Vector{UInt8}`.
  * `read(io, nbytes)` -> Reads the given number of bytes from the IO stream. Returns a `Vector{UInt8}` with `nbytes` number of elements containing the read contents.
  * `read(filename, String)` -> Returns the contents of the filename in a `String`.
  * `readline(io)` -> Reads the next line (terminated with `\n`) from the given stream into a `String` and returns it.
  * `readlines(io)` -> Reads all the lines at once and return a vector of strings.
  * `eachline(io)` -> Keeps reading the next line from the IO stream until it hits an EOF. Usually used in conjunction with a loop. By default this function will strip out the end-of-line but I can pass a kwarg `keep=true` to keep them.

> 💣 Don't try any of the `read` functions in a notebook, it will cause the notebook to hang. There is no text pop prompting for user input like in Python notebooks.

To read successive lines from stdin -

```julia
for line in eachline(stdin)
  print(line)
end
```

To read successive chars from stdin -

```julia
while !eof(stdin)
  x = read(stdin, Char)
  print(x)
end
```

## File Streams

For streaming content, I can open a filestream using the `open` function and close it after I am done with the `close` function. Just like in Python, Julia has convenience ways for the entire workflow of -

  1. Open the file descriptor
  2. Run a function that takes in the file descriptor as input and does stuff with it
  3. Close the file descriptor once the processing function returns

`open()` takes a function pointer as its first argument, which means I can use to `do-block` as shown in the example below.

````julia
# equivalent to open("duke_of_york.txt", "r")...
# default is read mode.
open("duke_of_york.txt") do f
    for line in eachline(f, keep=true)
        print("FOUND: $line")
    end
end
````

````julia
dukeofyork = """
Oh the grand old Duke of York
He had ten thousand men.
He led them up to the top of the hill
And he led them down again.
"""

open("poem.txt", "w") do f
    for line in readlines(IOBuffer(dukeofyork))
        write(f, string(line, '\n'))
    end
end
````
