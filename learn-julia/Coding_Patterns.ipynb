A pattern I see a lot with input validation is to use `&&` and `||` ops instead of if-conditions like so -

```julia
if cond
    action()
end
```

can be re-written as -

```julia
cond && action()
```

And 

```julia
if !cond
    action()
end
```

can be re-written as -

```julia
cond || action()
```

````julia
function positive(n::Integer)
    n > 0 && return nothing
    error("$n is not a positive number")
end

function bake(n::Integer)
    err = positive(n)
    err === nothing || throw(err)
    return 10n
end
````

````julia
bake(10)
````

````julia
try
    bake(-10)
catch err
    println(err)
end
````
