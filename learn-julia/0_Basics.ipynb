## Numeric Literal Co-efficients

Ref: https://docs.julialang.org/en/v1/manual/integers-and-floating-point-numbers/#man-numeric-literal-coefficients

The basic idea is that I can write expressions like $2x + 3$ just like in math.

````julia
f(x) = 2x + 3
````

````julia
f(2)
````

I can also write exponential functions or polynomials more elegantly than in other languages. $5x^2 + 3x + 1$ can be written as -

````julia
g(x) = 5x^2 + 3x + 1
````

````julia
g(2)
````

But when dealing with exponents there is a subtle rule I need to keep in mind. If I want to write $2x^3$, how do I write it? The intuitive answer is `2x^3`. But why doesn't this mean $(2x)^3$? When a numeric literal coefficient appears **after** an exponentiation operator (^), the coefficient acts as a complete unit for the exponent. But when it appears **before** an exponentiation, the exponentiation binds more tightly to its immediate operand.

##### Summary:

* `2^3x` → base is 2, exponent is 3x (the whole thing), mathematically it is $2^{3x}$
* `2x^3` → coefficient is 2, the x gets cubed, then multiply, mathematically it is $2x^3$

The intuitive answer is the correct one here.

## Comparison Operators

  * Value equality: `==` compares the values of two objects, they don't have to be exactly the same.
  * Identity equality : `===` compares the structures or memory location of the two objects. It is not just memory locations, because for immutable composite types, the memory location does not have to be the same.
  * Value inequality: `!=` or $\neq$ is the opposite of the value equality operator.
  * Identity inequality: `!==` is the opposite of identity equality.
  * Approximately equal: $\approx$ is shorthand for the `isapprox(x, y; atol, rtol)` function to compare floating point numbers.
  * Less than and greater than: `<=` or $\leq$ and `>=` or $\geq$ behave as expected.

````julia
ary1 = [1, 2, 3]
ary2 = [1, 2, 3]
````

````julia
ary1 == ary2
````

````julia
ary1 === ary2
````

````julia
5 == 5.0
````

````julia
5 === 5.0
````

````julia
1.21 ≈ 1.210000001
````

````julia
struct Cookie
    flavor::String
    calories::Int
end
````

````julia
c1 = Cookie("Chocolate Chip", 200)
c2 = Cookie("Chocolate Chip", 200)
````

````julia
c1 == c2
````

````julia
# Identity will also evaluate to true because Cookie is an immutable object
c1 === c2
````
