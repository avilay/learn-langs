# Binary to Decimal

Convert an int to a string representation of its binary digits.

```python
i = 8
bits = f"{i:b}"
print(type(bits), bits)
```

Convert a string representation of binary digits to int

```python
bits = "1011"
i = int(bits, 2)
print(type(i), i)
```

Convert int to bytes

```python
i = 8
bits = int(i).to_bytes(length=1, byteorder="big")
print(type(bits), bits)
```

Convert bytes to int

```python
bits = b"\x10"
i = int.from_bytes(bits, byteorder="big")
print(type(i), i)
```
